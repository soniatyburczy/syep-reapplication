# Code borough on the merged application frame

Applies the reviewed borough codebook (see `pipeline/codebook_validator.py`) to `data/processed/merged.csv` and writes `data/to_use/merged.csv`, the all-service-option lookup used to build the reapplication outcome.

Runs after `pipeline/merge_clean.py`.

```python
# Run from anywhere inside the repo: find the root, make it the working
# directory, and put it on sys.path so `pipeline/`, `descriptives/` and
# `paths.py` import cleanly. Data paths below are relative to the root.
import os, sys
from pathlib import Path

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'paths.py').exists())
os.chdir(ROOT)
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd
from pipeline.codebook_validator import CodebookConfig, apply_codebook
from pipeline.id_join import add_unique_id

df = pd.read_csv('data/processed/merged.csv', encoding='utf-8-sig')
```

```python
# Borough
borough_cfg = CodebookConfig(
    id_col="Participant.Unique.ID",
    raw_col="borough",
    codebook={},
    out_path="data/processed/borough_review.csv",
)

df = apply_codebook(df, borough_cfg, overwrite=True)
df['borough'] = df['borough'].replace('FALSE', 'Ambigious')
```

```python
df.to_csv('data/to_use/merged.csv', index=False)
```
